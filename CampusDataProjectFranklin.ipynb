### Cleaned Data Grouped by Recorder ID (First 7 Recorders Test)
We will extract the true Recorder ID (e.g., `001`, `002` from the original identifier column) and align it with the specific Location (e.g., `L01`, `L02`) and the target metrics.

```python
import pandas as pd
import re

# Create a fresh copy from the loaded df
mapping_df = pd.DataFrame()

# Extract clean Recorder ID (e.g. '001' from 'R26_001')
# The first column in 'df' contains the participant ID ('R26_001', etc.)
participant_col = df.columns[0]
mapping_df['Recorder_ID'] = df[participant_col].astype(str).str.extract(r'_(\d+)')

# Extract Location code (e.g. 'L01' from 'L01 MainEntrance')
location_col = df.columns[2]
mapping_df['Location'] = df[location_col].astype(str).str.extract(r'(L\d+)')

# Extract target metrics
temp_col = df.columns[4]    # Temperature
sound_col = df.columns[6]   # Sound level
urban_col = df.columns[8]   # Urban sound / traffic
trees_col = df.columns[10]  # Trees presence

mapping_df['Temperature_C'] = pd.to_numeric(df[temp_col], errors='coerce')
mapping_df['Sound_Level'] = df[sound_col]
mapping_df['Urban_Sound_Traffic'] = df[urban_col]
mapping_df['Trees_Presence'] = df[trees_col]

# Drop rows where critical mapping values are missing (like Recorder_ID or Location)
mapping_df = mapping_df.dropna(subset=['Recorder_ID', 'Location'])

# Filter for the first 7 Recorder IDs as a test ('001' to '007')
test_recorders = [f"{i:03d}" for i in range(1, 8)]
test_df = mapping_df[mapping_df['Recorder_ID'].isin(test_recorders)].reset_index(drop=True)

# Display the structured DataFrame
display(test_df)
```

### Cleaned Data Grouped by Recorder ID (Next 7 Recorders: 008 to 014)
This section applies the same structured mapping format for the next set of recorder IDs.

```python
# Filter for the next 7 Recorder IDs ('008' to '014')
next_recorders = [f"{i:03d}" for i in range(8, 15)]
next_df = mapping_df[mapping_df['Recorder_ID'].isin(next_recorders)].reset_index(drop=True)

# Display the structured DataFrame
display(next_df)
```
